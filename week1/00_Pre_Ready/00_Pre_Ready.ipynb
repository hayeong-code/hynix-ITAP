## 3. Anaconda 설치 — 자신의 운영체제만 진행

### Windows

1. 내려받은 `.exe` 파일을 더블클릭합니다.
2. 약관을 읽고 동의한다면 다음 단계로 진행합니다.
3. 설치 대상은 개인 사용자용 `Just Me`를 선택합니다.
4. 설치 위치를 확인합니다. 가능하면 공백·특수문자가 없는 경로를 사용합니다.
5. `Add Anaconda to my PATH` 옵션은 선택하지 않습니다. 기존 Python을 기본 프로그램으로 바꾸는 옵션은 기존 환경이 있다면 해제합니다.
6. 설치가 완료되면 시작 메뉴에서 **Anaconda Prompt**를 검색해 실행합니다. 아래 명령은 일반 Python 창이 아니라 이 창에 입력합니다.

설치 화면의 명칭은 버전에 따라 조금 다를 수 있습니다. [공식 Windows 설치 안내](https://www.anaconda.com/docs/getting-started/anaconda/install/windows-gui-install), [PATH와 환경 활성화 안내](https://docs.conda.io/projects/conda/en/stable/user-guide/tasks/manage-environments.html)

### macOS — Apple Silicon

1. Apple 메뉴의 **이 Mac에 관하여**에서 칩 종류를 확인합니다.
2. Apple Silicon용 `.pkg` 파일을 열고 약관·설치 위치를 확인하며 진행합니다.
3. 설치가 끝나면 **터미널(Terminal)**을 새로 엽니다.
4. 아래 `conda --version` 명령으로 설치를 확인합니다.

공식 GUI 설치의 기본 경로는 `/opt/anaconda3`입니다. 설치 위치를 바꿨다면 실제 경로를 사용하세요. [공식 macOS 설치 안내](https://www.anaconda.com/docs/getting-started/anaconda/install/mac-gui-install)

### Linux

1. 공식 다운로드에서 CPU에 맞는 `.sh` 파일을 받습니다.
2. 터미널에서 설치 파일을 실행합니다. 아래 따옴표 안 경로는 **실제로 내려받은 파일 경로**로 바꾸세요.

```bash
bash "다운로드한_설치파일의_전체경로.sh"
```

3. 안내에 따라 약관과 설치 위치를 확인합니다.
4. conda 초기화 여부를 물으면 `yes`를 선택합니다. 이후 터미널에서 conda를 사용할 수 있게 하는 설정입니다.
5. 설치 후 터미널을 닫고 새로 엽니다. [공식 Linux 설치 안내](https://www.anaconda.com/docs/getting-started/anaconda/install/linux-install)

### 공통 설치 확인

**입력 위치: Windows는 Anaconda Prompt, macOS/Linux는 터미널**

```text
conda --version
```

`conda 26.x.x`처럼 버전이 표시되면 다음 단계로 갑니다. 정확히 같은 버전일 필요는 없습니다.
`>>>`가 보이면 Python 대화형 창입니다. `exit()`로 나온 뒤 명령을 실행하세요. `(base)` 같은 문구는 화면 표시이므로 직접 입력하지 않습니다.

## 4. 수업용 Python 환경 만들기

환경은 수업에 필요한 Python과 라이브러리를 모아두는 독립된 작업 공간입니다. 기본 `base`와 구분해 `hynix-eda`를 만듭니다.

**입력 위치: Anaconda Prompt / 터미널. 각 줄을 입력하고 Enter를 누르세요.**

```text
conda create -n hynix-eda python=3.12 pip
conda activate hynix-eda
python --version
python -c "import sys; print(sys.executable)"
```

첫 명령에서 설치할 패키지 목록을 확인하고 진행 여부에 `y`를 입력합니다. 다운로드가 끝난 뒤 다음 줄을 실행하세요. 마지막 출력에서 Python 3.12와 `hynix-eda` 환경의 경로를 확인합니다. 이미 같은 환경을 만들었다면 `conda create`를 반복하지 않고 `conda activate`부터 진행합니다. [conda 환경 생성·활성화 안내](https://docs.conda.io/projects/conda/en/latest/user-guide/getting-started.html)

## 5. 교재 폴더로 이동하고 라이브러리 설치하기

### 5-1. 폴더 이동

`cd`는 작업 폴더를 바꾸는 명령입니다. 아래는 예시 경로이므로 **자신이 교재를 저장한 위치**로 수정합니다. 파일 탐색기/Finder에서 `requirements.txt`가 있는 폴더의 경로를 확인하세요.

Windows Anaconda Prompt:

```bat
cd /d "C:\training\hynix-iTAP"
dir requirements.txt
```

macOS/Linux 터미널:

```bash
cd "$HOME/training/hynix-iTAP"
ls requirements.txt
```

`requirements.txt`가 출력되면 올바른 위치입니다. 파일이 없다고 나오면 설치를 계속하기 전에 경로를 수정하세요.

### 5-2. 필요한 패키지 설치

같은 창에서 `(hynix-eda)` 환경이 활성화되어 있는지 확인한 뒤 실행합니다.

```text
python -m pip install -r requirements.txt
python -c "import numpy, pandas, matplotlib, jupyterlab, ipykernel; print('Library check OK')"
```

`requirements.txt`에는 NumPy·Pandas·Matplotlib·JupyterLab 등 수업에 필요한 패키지가 적혀 있습니다. `python -m pip`는 지금 선택한 Python 환경에 설치하는 명령입니다. 마지막에 `Library check OK`가 나오면 라이브러리 import가 성공한 것입니다.

### 5-3. Notebook에서 선택할 Kernel 등록

```text
python -m ipykernel install --user --name hynix-eda --display-name "Python (hynix-eda)"
```

이 명령은 Jupyter의 Kernel 선택 목록에 **Python (hynix-eda)**를 추가합니다. 환경 생성과 Kernel 등록은 처음 한 번 하면 됩니다. 같은 이름으로 다시 등록하면 해당 Kernel 설정이 갱신됩니다. [IPython 공식 Kernel 등록 안내](https://ipython.readthedocs.io/en/stable/install/kernel_install.html)

## 8. 자주 만나는 문제

| 증상 | 먼저 확인할 것 |
|---|---|
| `conda`를 찾을 수 없음 | Windows는 Anaconda Prompt인지 확인. macOS/Linux는 설치 후 터미널을 다시 열었는지 확인 |
| macOS/Linux에서 여전히 conda를 못 찾음 | 실제 설치 경로의 `bin/activate`를 `source`로 실행. 기본 GUI macOS 예: `source /opt/anaconda3/bin/activate`, 기본 Linux 예: `source ~/anaconda3/bin/activate` |
| `hynix-eda` 환경이 이미 존재함 | `conda env list`로 확인한 뒤 `conda activate hynix-eda` 실행 |
| `requirements.txt`를 못 찾음 | ZIP 압축을 풀었는지, 현재 위치가 프로젝트 루트인지 확인 |
| `ModuleNotFoundError` | Kernel이 Python (hynix-eda)인지 확인. 해당 환경을 활성화한 터미널에서 패키지 설치 후 Kernel 재시작 |
| `Python (hynix-eda)`가 목록에 없음 | 5-3 Kernel 등록을 실행하고 JupyterLab을 새로고침 |
| `SyntaxError`가 나며 conda/pip 줄을 가리킴 | 터미널 명령을 Notebook Code 셀에 입력하지 않았는지 확인 |
| CSV가 없다고 나옴 | 프로젝트 루트의 `data`를 포함해 교재 전체를 받았는지 확인 |
| `NameError` | 필요한 변수를 만드는 앞 셀부터 실행하거나 Kernel 재시작 후 전체 실행 |
| 설치 중 네트워크·인증서 오류 | 오류 문구를 기록해 강사/IT 담당자에게 공유하고 승인된 네트워크·패키지 서버 설정 확인 |

화면이 다르면 오류 문구와 함께 사용한 OS, 실행한 명령, `python --version` 및 `sys.executable` 출력을 전달하면 원인을 찾는 데 도움이 됩니다.

별도 **Jupyter Notebook 앱**을 사용할 때만, 활성화한 `hynix-eda` 환경의 터미널에서 다음을 실행합니다. JupyterLab으로 실습한다면 추가 설치할 필요가 없습니다.

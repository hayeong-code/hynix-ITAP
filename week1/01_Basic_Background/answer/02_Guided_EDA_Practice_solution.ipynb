## Q4. 어떤 열에서 얼마나 기록이 빠졌나요?

**앞 단계와 연결:** Q2의 non-null count 차이를 결측 개수와 비율로 정리합니다.

**할 일:** 열별 결측 개수와 결측률(%)을 한 표로 만들고 결측률이 큰 순서로 정렬하세요. 결과를 missing_summary에 저장하세요.

**Hint:** `isnull()`, `sum()`, 전체 행 수, `pd.DataFrame()`, `sort_values()`. 결측률 = 결측 개수 ÷ 전체 행 수 × 100.

## Q18. [심화 ★★] 품질 이상치 후보를 일관된 규칙으로 표시할 수 있나요?

**앞 단계와 연결:** Q9에서 눈으로 찾은 극단값을 IQR 규칙으로 표시합니다. 삭제 여부는 아직 결정하지 않습니다.

**할 일:** Q9의 quality_values에서 Q1(25%), Q3(75%), IQR을 구하세요. [Q1 − 1.5×IQR, Q3 + 1.5×IQR] 밖의 품질값을 quality_outlier_mask로 표시하고 경계·후보 수·해당 행을 출력하세요. 결측은 후보로 세지 마세요.

**Hint:** `quantile()`, 조건 두 개를 `|`로 연결. IQR = Q3 − Q1은 가운데 50% 구간의 폭입니다. 여기서 Q1·Q3는 사분위수를 뜻하며 문제 번호가 아닙니다.

```python
quality_q1 = quality_values.quantile(0.25)
quality_q3 = quality_values.quantile(0.75)
quality_iqr = quality_q3 - quality_q1
lower_bound = quality_q1 - 1.5 * quality_iqr
upper_bound = quality_q3 + 1.5 * quality_iqr
quality_outlier_mask = df["QUALITY_VALUE"].notna() & (
    (df["QUALITY_VALUE"] < lower_bound) | (df["QUALITY_VALUE"] > upper_bound)
)
print("하한 / 상한:", lower_bound, upper_bound)
print("후보 수:", quality_outlier_mask.sum())
display(df.loc[quality_outlier_mask].sort_values("QUALITY_VALUE"))
```

**결과 확인:** 전체 데이터의 경계와 그룹별 경계는 같을까요? 이 후보 수가 생성 코드에서 넣은 극단값 수와 반드시 같아야 하나요?

**할 일:** 품질값이 유효하면서 IQR 후보가 아닌 행만 sensitivity_df로 만드세요. 장비별 mean·median·std·count를 다시 계산해 Q13 표와 나란히 비교하고 mean 변화(After − Before), 유효 count 감소를 계산하세요. 원본 df는 유지하세요.

**Hint:** `notna()`와 후보 mask의 반대(`~`)를 결합합니다. `.join(..., lsuffix=..., rsuffix=...)`으로 전후 열을 구분할 수 있습니다. Before의 count도 유효값 기준이어야 합니다.

```python
sensitivity_df = df.loc[
    df["QUALITY_VALUE"].notna() & ~quality_outlier_mask
].copy()
after_equipment_summary = sensitivity_df.groupby("EQUIPMENT")["QUALITY_VALUE"].agg(
    ["mean", "median", "std", "count"]
)
sensitivity_comparison = equipment_summary.join(
    after_equipment_summary, lsuffix="_before", rsuffix="_after"
)
```

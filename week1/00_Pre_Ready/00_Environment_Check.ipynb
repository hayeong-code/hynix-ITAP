## 5. 실습 CSV 경로 확인

교재 폴더 전체가 준비되었는지 확인합니다. 아래는 경로 확인용 코드이므로 그대로 실행하세요.
`(400, 8)`이 출력되어야 합니다. 이 파일은 **제조 공정을 모사한 합성 데이터**이며 실제 기업 데이터가 아닙니다.

```python
from pathlib import Path

# 실행 위치와 관계없이 requirements.txt와 week1이 있는 프로젝트 루트를 찾습니다.
current_dir = Path.cwd()
project_root = next(
    (folder for folder in [current_dir, *current_dir.parents]
     if (folder / "requirements.txt").is_file() and (folder / "week1").is_dir()),
    None,
)
if project_root is None:
    raise FileNotFoundError("교재 폴더 전체를 내려받고 hynix-iTAP 안에서 실행하세요.")

csv_path = project_root / "data/practice_process_data.csv"
practice_df = pd.read_csv(csv_path)
print("파일:", csv_path.name)
print("데이터 크기:", practice_df.shape)
print("400행 × 8열 확인:", practice_df.shape == (400, 8))
```

## 6. 그래프 출력 확인

Coffee·Tea의 주문 수량 2·3을 나타내는 막대 두 개가 표시되어야 합니다.
제목과 축 이름은 한글 폰트 설치 없이 확인할 수 있도록 영어로 표시합니다.

```python
plt.figure(figsize=(5, 3))
plt.bar(check_df["MENU"], check_df["QUANTITY"])
plt.title("Environment check: order quantity")
plt.xlabel("MENU")
plt.ylabel("QUANTITY (cups)")
plt.tight_layout()
plt.show()
```
